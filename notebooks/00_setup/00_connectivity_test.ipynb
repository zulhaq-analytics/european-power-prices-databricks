```python
import requests, socket

targets = {
    "ENTSO-E API":   "https://web-api.tp.entsoe.eu/api",
    "ENTSO-E site":  "https://transparency.entsoe.eu",
    "GitHub raw":    "https://raw.githubusercontent.com/github/gitignore/main/Python.gitignore",
    "PyPI":          "https://pypi.org/simple/requests/",
    "Google":        "https://www.google.com",
}

print(f"{'Target':<14} | {'DNS':<16} | {'Result'}")
print("-" * 80)
for name, url in targets.items():
    host = url.split("/")[2]
    try:
        ip = socket.gethostbyname(host)
    except Exception as e:
        ip = f"FAIL ({type(e).__name__})"
    try:
        r = requests.get(url, timeout=15)
        body = r.text[:120].replace("\n", " ")
        result = f"HTTP {r.status_code} | {body}"
    except Exception as e:
        result = f"BLOCKED/ERROR: {type(e).__name__}: {str(e)[:120]}"
    print(f"{name:<14} | {ip:<16} | {result}")
```
